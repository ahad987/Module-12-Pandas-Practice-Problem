# 1.	Create a Pandas Series from the list [10, 20, 30, 40, 50]. Print the Series, its index, and data type. Then create a DataFrame with columns "Name", "Age", and "City" containing at least 4 rows of data.

```python
import pandas as pd
```

```python
# Create a Python List
numbers = [10, 20, 30, 40, 50]

# Convert List into Pandas Series
my_series = pd.Series(numbers)

# Print the Series
print("Pandas Series:")
print(my_series)

# Print Index
print("\nSeries Index:")
print(my_series.index)

# Print Data Type
print("\nData Type:")
print(my_series.dtype)
```

# 2.	Using any DataFrame, write code to print the number of rows and columns, column data types (.dtypes), and summary statistics (.describe()).

```python
# Create DataFrame
data = {
    "Name": ["Rahim", "Karim", "Sakib", "Nusrat", "Ayesha"],
    "Age": [22, 25, 23, 24, 26],
    "Salary": [20000, 25000, 22000, 28000, 30000],
    "Score": [75, 85, 65, 90, 80]
}

df = pd.DataFrame(data)

# Print DataFrame
print("Original DataFrame:")
print(df)

# Number of Rows and Columns
print("\nNumber of Rows and Columns:")
print(df.shape)

# Number of Rows
print("\nNumber of Rows:")
print(df.shape[0])

# Number of Columns
print("\nNumber of Columns:")
print(df.shape[1])

# Column Data Types
print("\nData Types:")
print(df.dtypes)

# Summary Statistics
print("\nSummary Statistics:")
print(df.describe())
```

# 4. Using .loc[] and .iloc[], show the first row, the second column, and a 2×2 block    from the middle of your DataFrame.

```python
data = {
    "Product": ["Pen", "Book", "Eraser", "Pencil"],
    "Price": [10, 50, 5, 7],
    "Quantity": [100, 60, 200, 150]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
```

# **Part A: Using .loc[]**

```python
print(df.loc[0])
print("\nPrice-------------------")
print(df.loc[:, "Price"])
print("\nPrice and Quantity------")
print(df.loc[1:2, ["Price", "Quantity"]])
```

# **Part B: Using .iloc[]**

```python
print(df.iloc[0])

print("\nPrice-------------------")
print(df.iloc[:, 1])

print("\nPrice and Quantity------")
print(df.iloc[1:3, 1:3])
```

# 5. Create a 4×3 DataFrame of student names and marks. Drop one column using .drop(), drop one row by index.

```python
# Create DataFrame
data = {
    "Name": ["Rahim", "Karim", "Sakib", "Nusrat"],
    "Math": [85, 75, 90, 80],
    "English": [70, 80, 85, 90]
}

df = pd.DataFrame(data)

# Print Original DataFrame
print("Original DataFrame:")
print(df)

# Drop one Column
df_drop_column = df.drop("Math", axis=1)

print("\nAfter Dropping Math Column:")
print(df_drop_column)

# Drop one Row using Index
df_drop_row = df.drop(1, axis=0)

print("\nAfter Dropping Row with Index 1:")
print(df_drop_row)
```

# 6.	Create a DataFrame of employees with columns "Name", "Department", and "Salary". Sort it by "Salary" in ascending order and by "Name" in descending order.

```python
# Create Employee DataFrame
data = {
    "Name": ["Rahim", "Karim", "Sakib", "Nusrat", "Ayesha"],
    "Department": [
        "IT",
        "HR",
        "Finance",
        "IT",
        "Marketing"
    ],
    "Salary": [35000, 25000, 45000, 30000, 40000]
}

df = pd.DataFrame(data)

# Print Original DataFrame
print("Original Employee DataFrame:")
print(df)

# Sort by Salary in Ascending Order
salary_sorted = df.sort_values(
    by="Salary",
    ascending=True
)

print("\nSorted by Salary (Ascending):")
print(salary_sorted)

# Sort by Name in Descending Order
name_sorted = df.sort_values(
    by="Name",
    ascending=False
)

print("\nSorted by Name (Descending):")
print(name_sorted)

# Sort by Department in Descending Order and Salary in Ascending Order
both_sorted = df.sort_values(
    by=["Department", "Salary"],
    ascending=[False, True]
)

print("\nSorted by Department (Descending): and Salary (Ascending)")
print(both_sorted)
```

# 7.	Create a DataFrame of 8 students with columns "Name" and "Score". Filter and print students with Score > 60 and those with Score < 40.

```python
# Create Student DataFrame
data = {
    "Name": [
        "Rahim",
        "Karim",
        "Sakib",
        "Nusrat",
        "Ayesha",
        "Hasan",
        "Mim",
        "Tanvir"
    ],

    "Score": [85, 35, 75, 90, 25, 55, 65, 30]
}

df = pd.DataFrame(data)

# Print Original DataFrame
print("Original Student DataFrame:")
print(df)

# Filter Students with Score > 60
high_score = df[df["Score"] > 60]

print("\nStudents with Score > 60:")
print(high_score)

# Filter Students with Score < 40
low_score = df[df["Score"] < 40]

print("\nStudents with Score < 40:")
print(low_score)
```

```python
# Filter Students with Score with 40 to 80

result = df[
    (df["Score"] >= 40) &
    (df["Score"] <= 80)
]

print(result)
```
